# Add Top-of-File Role Comments
This notebook demonstrates how to inject top-of-file role comments into project files and verify the headers were added.

```python
from pathlib import Path

# Define the repository root and sample file definitions
repo_root = Path("d:/RAG/CollegeProject")

sample_files = {
    "travel_agents/nodes.py": "Routes conversation turns into agent nodes and handles booking, planning, research, and small talk.",
    "travel_tools/search_engine.py": "Loads resort metadata, builds search indexes, and retrieves matching resort records.",
    "frontend/components/Navbar.jsx": "Navigation bar component used across frontend pages.",
    "README.md": "Project overview, setup instructions, and general documentation for the travel assistant.",
}

print(f"Repository root: {repo_root}")
print("Defined sample files for header injection:")
for path, comment in sample_files.items():
    print(f" - {path}: {comment}")
```

## Add Top-of-File Role Comments
Create sample files and insert a top-of-file role comment into each one.

```python
created_files = []
for rel_path, comment in sample_files.items():
    file_path = repo_root / rel_path
    file_path.parent.mkdir(parents=True, exist_ok=True)
    content = "print('Hello')\n" if file_path.suffix == ".py" else "export default function Component() { return null; }\n" if file_path.suffix in {".jsx"} else "# README placeholder\n"
    header = {
        ".py": f"# {comment}\n# File: {rel_path}\n\n",
        ".jsx": f"/* {comment} */\n/* File: {rel_path} */\n\n",
        ".md": f"<!-- {comment} -->\n<!-- File: {rel_path} -->\n\n",
    }[file_path.suffix]
    file_path.write_text(header + content, encoding="utf-8")
    created_files.append(file_path)

print("Created sample files with headers:")
for path in created_files:
    print(f" - {path}")
```

## Read and Display File Headers
Read the top lines from each sample file to verify that the header comments were added correctly.

```python
for rel_path in sample_files:
    file_path = repo_root / rel_path
    print(f"\n=== {rel_path} ===")
    for line in file_path.read_text(encoding='utf-8').splitlines()[:4]:
        print(line)
```

## Automate Comment Injection
Define a reusable function that injects or updates top-of-file role comments for multiple files in the project.

```python
def inject_file_header(file_path: Path, comment: str):
    text = file_path.read_text(encoding="utf-8")
    lines = text.splitlines()
    suffix = file_path.suffix

    if suffix == ".py":
        if len(lines) >= 2 and lines[0].startswith("# ") and lines[1].startswith("# File:"):
            text = "\n".join(lines[2:])
    elif suffix in {".js", ".jsx", ".mjs"}:
        if len(lines) >= 2 and lines[0].startswith("/* ") and lines[1].startswith("/* File:"):
            text = "\n".join(lines[2:])
    elif suffix == ".md":
        if len(lines) >= 2 and lines[0].startswith("<!-- ") and lines[1].startswith("<!-- File:"):
            text = "\n".join(lines[2:])
    else:
        raise ValueError(f"Unsupported suffix: {suffix}")

    rel_path = file_path.relative_to(repo_root).as_posix()
    if suffix == ".py":
        header = f"# {comment}\n# File: {rel_path}\n\n"
    elif suffix in {".js", ".jsx", ".mjs"}:
        header = f"/* {comment} */\n/* File: {rel_path} */\n\n"
    elif suffix == ".md":
        header = f"<!-- {comment} -->\n<!-- File: {rel_path} -->\n\n"

    file_path.write_text(header + text, encoding="utf-8")
    print(f"Injected header into {file_path}")

file_comments = {
    "travel_tools/__init__.py": "Marks travel_tools as a Python package for booking/search utilities.",
    "travel_tools/vectorstore_provider.py": "Provides vector store integration for similarity search and embedding lookups.",
    "travel_tools/search_tool.py": "Formats resort search results and fallback responses for the LLM-driven travel assistant.",
    "travel_tools/search_engine.py": "Loads resort metadata, builds search indexes, and retrieves matching resort records.",
    "travel_tools/data_validator.py": "Validates travel and booking request inputs before they are processed.",
    "travel_tools/config.py": "Holds configuration values used by travel tools and backend services.",
    "travel_tools/booking_tool.py": "Handles booking request creation and submission to the travel booking API.",
    "travel_tools/booking_store.py": "Stores and retrieves booking records for the reservation workflow.",
    "travel_tools/booking_status_tool.py": "Provides booking status lookup functionality for user inquiries.",
    "tests/__init__.py": "Marks the tests directory as a Python package.",
    "tests/test_search_tool.py": "Tests the resort search tool formatting and result handling.",
    "tests/test_search_engine.py": "Tests resort search engine behavior and matching logic.",
    "tests/test_nodes.py": "Tests travel agent routing and booking node behavior.",
    "tests/test_booking_tool.py": "Tests booking request creation and booking tool integration.",
    "tests/test_basic.py": "Provides basic sanity checks for the travel assistant application.",
    "travel_api/__init__.py": "Marks the travel_api backend as a Python package.",
    "travel_api/store.py": "Implements persistent storage access for resorts, sessions, and bookings.",
    "travel_api/services.py": "Defines backend service logic for travel booking and resort data operations.",
    "travel_api/models.py": "Defines data models used by the travel API and booking system.",
    "travel_api/app.py": "Defines the main travel API application and HTTP routing.",
    "test_resort_search.py": "Ad hoc test or script for resort search validation outside the main test suite.",
    "tmp_test_parser.py": "Temporary parser proof-of-concept used for validating date parsing logic.",
    "run_tests.py": "Entrypoint script for running the project's Python test suite.",
    "tests/conftest.py": "Configures shared pytest fixtures and test setup helpers.",
    "tests/integration/test_api_endpoints.py": "Integration tests for travel API endpoints and request flows.",
    "tests/integration/__init__.py": "Marks integration tests as a Python package.",
    "travel_agents/__init__.py": "Marks the travel_agents package for conversational agent state and routing.",
    "travel_agents/state.py": "Tracks conversational booking state and context for the travel assistant.",
    "travel_agents/resort_helpers.py": "Resolves resort follow-up queries and assists with resort-specific answers.",
    "travel_agents/nodes.py": "Routes conversation turns into agent nodes and handles booking, planning, research, and small talk.",
    "travel_agents/llms.py": "Configures LLM clients and prompt chains used by the travel assistant.",
    "travel_agents/intent.py": "Analyzes user intent and maps phrases to booking and search actions.",
    "travel_agents/graph.py": "Builds graph structures or flows used by the travel agent.",
    "travel_agents/date_helpers.py": "Provides date parsing and normalization utilities for booking workflows.",
    "travel_agents/content.py": "Extracts and sanitizes text content from messages and assistant responses.",
    "travel_agents/booking_helpers.py": "Extracts booking-specific context and helpers used during booking conversations.",
    "frontend/next.config.mjs": "Next.js configuration for the frontend application.",
    "frontend/middleware.js": "Defines frontend middleware for request handling and routing behavior.",
    "frontend/lib/useScrollAnimation.js": "Custom hook for scroll-based animation effects in the frontend.",
    "frontend/components/Wildlife/TiltCard.jsx": "UI component for animated wildlife-themed tilt cards.",
    "frontend/components/Wildlife/TigerGuide.jsx": "Wildlife-themed UI component displaying tiger guide content.",
    "frontend/components/Wildlife/Butterflies.jsx": "UI component rendering butterfly-themed animation visuals.",
    "frontend/components/Wildlife/AmbientBees.jsx": "UI component rendering ambient bee animation effects.",
    "frontend/components/Skeletons.jsx": "UI component providing skeleton loading placeholders.",
    "frontend/components/Sidebar.jsx": "Sidebar navigation component for the frontend UI.",
    "frontend/components/ScrollTrigger.jsx": "Scroll-triggered animation helper component.",
    "frontend/components/ScrollAnimation.jsx": "Component for applying scroll-based animations on page sections.",
    "frontend/components/Navbar.jsx": "Navigation bar component used across frontend pages.",
    "frontend/components/MessageThread.jsx": "Chat UI component displaying conversation threads.",
    "frontend/components/Footer.jsx": "Footer component for the frontend site.",
    "frontend/components/Composer.jsx": "Chat input component for composing new user messages.",
    "frontend/components/ClerkAuthPanel.jsx": "Authentication UI panel integrating Clerk sign-in/sign-up flows.",
    "frontend/lib/auth.js": "Frontend authentication utilities and session helpers.",
    "frontend/lib/api.js": "Client-side API helper functions for the frontend.",
    "frontend/lib/server/mongodb.js": "Server-side MongoDB helper for persistent session storage.",
    "frontend/lib/server/fileSessionStore.js": "Fallback server-side file-based session store implementation.",
    "frontend/lib/server/pythonApi.js": "Client wrapper for calling the Python backend API from the frontend.",
    "frontend/lib/server/sessionStore.js": "Server-side session storage utilities for the frontend application.",
    "frontend/app/layout.jsx": "Root layout for the Next.js frontend application.",
    "frontend/app/sign-up/page.jsx": "Sign-up page for new users in the frontend.",
    "frontend/app/(marketing)/page.jsx": "Marketing home page for the travel assistant frontend.",
    "frontend/app/(marketing)/layout.jsx": "Layout wrapper for marketing pages in the frontend.",
    "frontend/app/chat/page.jsx": "Main chat page for interacting with the travel assistant bot.",
    "frontend/app/sign-in/page.jsx": "Sign-in page for returning users.",
    "frontend/app/(marketing)/how-it-works/page.jsx": "Marketing page explaining how the travel assistant works.",
    "frontend/app/(marketing)/about/page.jsx": "Marketing page with information about the travel service.",
    "frontend/app/(marketing)/explore/page.jsx": "Marketing page showcasing resort and destination exploration.",
    "frontend/app/api/sessions/route.js": "API route for storing and retrieving chat session state.",
    "frontend/app/api/config/route.js": "API route exposing frontend configuration values.",
    "frontend/app/api/chat/stream/route.js": "Streaming chat API route for conversational messages.",
    "frontend/app/api/sessions/[sessionId]/route.js": "API route for session-specific session retrieval and updates.",
    "README.md": "Project overview, setup instructions, and general documentation for the travel assistant.",
    "RESORT_ISSUE_ANALYSIS.md": "Document describing the resort data issue analysis and troubleshooting findings.",
    "RESORT_FIX_SUMMARY.md": "Summary of resort fixes applied in the project.",
    "PROJECT_SUMMARY_FOR_PRESENTATION.md": "Project summary content prepared for presentation use.",
    "PRESENTATION_ACTION_PLAN.md": "Action plan for project presentation and next steps.",
    "RENDER_SETUP.md": "Documentation for local rendering setup and environment configuration.",
    "RENDER_FIX_NOW.md": "Guide for immediate render fixes and troubleshooting.",
    "RESPONSIVE_DESIGN.md": "Guide describing responsive design considerations for the project.",
    "RESPONSIVE_COMPONENTS_GUIDE.md": "Guide for responsive frontend components and layout patterns.",
    "RESPONSIVE_CHECKLIST.md": "Responsive design checklist for development and testing.",
    "API_ROTATION_SYSTEM.md": "Documentation for the API rotation system used by the project.",
    "TESTING_SETUP_GUIDE.md": "Guide for setting up tests and test automation.",
    "SCROLL_ANIMATIONS_GUIDE.md": "Guide for scroll animation implementation and behavior.",
    "DUAL_API_ROTATION_SYSTEM.md": "Documentation for the dual API rotation system design.",
    "FINAL_YEAR_PROJECT_CHECKLIST.md": "Final-year project checklist for milestones and deliverables.",
}

for rel_path, comment in file_comments.items():
    inject_file_header(repo_root / rel_path, comment)
```
